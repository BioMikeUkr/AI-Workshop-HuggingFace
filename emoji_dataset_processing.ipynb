# 2. Preparing and resizing images for the dataset
In this step, we process all images by resizing them to 32×32 pixels and prepare them for upload as a Hugging Face dataset. We:

- Read all images from the emoji_dataset_images folder,

- Resize each image to 32×32 and save it into a new folder emoji_resized,

- Build a Hugging Face Dataset using datasets.Image so that images are recognized and displayed properly in the Hub viewer.

```python
from datasets import Dataset, DatasetDict, Features, Image as HFImage
from PIL import Image
import os

data_dir = "emoji_dataset_images"
output_dir = "emoji_resized"

os.makedirs(output_dir, exist_ok=True)

file_list = sorted([f for f in os.listdir(data_dir) if f.endswith(('.png', '.jpg', '.jpeg'))])

images = []
for file in file_list:
    input_path = os.path.join(data_dir, file)
    output_path = os.path.join(output_dir, file)
    img = Image.open(input_path).convert("RGB").resize((32, 32))
    img.save(output_path)
    images.append({"image": output_path})

features = Features({
    "image": HFImage()
})

dataset = Dataset.from_list(images, features=features)
dataset_dict = DatasetDict({"train": dataset})
```
